```python
###A fetch that never crashes
```

```python
def get_rates(base):
    try:
        response = requests.get("https://api.frankfurter.dev/v1/latest",
                               params={"from": base})
        if response.ok:
            return response.json()["rates"]
        else:
            return f"Error: API returned {response.status_code} - {response.text}"
    except Exception as error:
        return f"Error: {error}"
```

```python
currencies_to_test = ["ÜSD","EUR","GBP","JPY","CHF","NGN","ZWL","XYZ","ng"]
for currency in currencies_to_test:
    print("\n", currency)
    print(get_rates(currency))
```

```python
### Users into a dictionary
```

```python
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
users = response.json()
users_dict = {}
for user in users:
    users_dict[user["id"]] = {"name": user["name"], "email": user["email"], "city": user["address"]["city"]}
print(users_dict)
```
